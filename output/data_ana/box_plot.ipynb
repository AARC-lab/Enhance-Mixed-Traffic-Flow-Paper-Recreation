```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('/home/ruby/Nazmus Shakib/AARC Lab/Paper Recreation: Enhance Mix Traffic Flow/output/data/delay_data/delay_mpr_40_flow_54_12_time_3600_ramp.csv')
```

```python
data['vehicle_type'] = data['vehicle_id'].apply(lambda x: 'HDV' if 'HDV' in x else 'CAV')
plt.figure(figsize=(8, 6))
sns.boxplot(x='vehicle_type', y='delay_time', data=data)
plt.title('Delay Time by Vehicle Type')
plt.xlabel('Vehicle Type')
plt.ylabel('Delay Time')
plt.show()
```
